```python
# Part 1
def f(x): # Polynomial function
    """
    This function calculates the polynomial function of x. 
    Parameter x  -- an integer or float number;
    Return: f(x) -- an integer or float number.
    """
    return x**3 - x**2 - 1

def df(x): # Derivative of Polynomial function
    """
    This function calculates the derivative of polynomial function of x. 
    Parameter x: an integer or float number;
    Return: df/dx: an integer or float number.
    """
    return 3*x**2 - 2*x
```

```python
# Example test
f(2) # expected value of 3
```

```python
# Example test
df(2) # expected value of 8
```

```python
# Part 2
def newton(f, df, x0, epsilon=1e-6, max_iter=30):
    """
    This function performs the newton iteration of function f and df.
    Parameter:
        f  -- func, polynomial func;
        df -- func, derivative of polynomial func;
        x0 -- int or float, initial guess for the root of the function;
        epsilon -- float, the precision threshold of the evaluated value, 
        the absolute value of evaluated and expected value needs 
        to be smaller and equal to epsilon
        max_iter -- int, the maximum iteration that the function could run
        
    Return: the value of x_n such that f(x_n) = 0, 
    as well as the iteration turns the newton function ran to find x_n.
    Otherwise, return iternation failed.
    """
    
    x = x0
    n = 0
    while n < max_iter:
        if abs(f(x)) < epsilon:
            print(f"Found root in {n} iterations")
            return x
        
        x = x - f(x) / df(x)
        n = n + 1
        
    
    print("iteration failed")
    return None
```

```python
# Part 3.1
newton(f, df, 1.64)
```

```python
# Part 3.2
newton(f, df, 3)
```
